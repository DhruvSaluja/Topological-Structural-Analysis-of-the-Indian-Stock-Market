```python
import pandas as pd
import numpy as np

all_stocks = []

for i, filename in enumerate(allcompanyfiles):

    df = pd.read_csv("./Dataset/" + filename)

    # Extract dates and closing prices
    dates = pd.to_datetime(df.iloc[2:, 0])

    prices = pd.to_numeric(df.iloc[2:, 1])

    # Create date-indexed series
    stock = pd.Series(
        prices.values,
        index=dates,
        name=f"{filename}"
    )

    all_stocks.append(stock)

# Align all companies using dates
stock_data = pd.concat(all_stocks, axis=1)

# Sort dates
stock_data = stock_data.sort_index()

print(stock_data.shape)
print(stock_data.head())
stock_data
```

```python
import numpy as np
import pandas as pd

# stock_data shape: (246 dates, 200 stocks)

# Calculate logarithmic returns
log_returns = np.log(stock_data / stock_data.shift(1))

print(log_returns.shape)
log_returns
```

```python
correlation_matrix_logreturn = log_returns.corr(method='pearson')

print(correlation_matrix_logreturn.shape)
correlation_matrix_logreturn
```

```python
correlation_matrix_closeprices = stock_data.corr(method='pearson')

print(correlation_matrix_closeprices.shape)
correlation_matrix_closeprices
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 12))

sns.heatmap(
    correlation_matrix_logreturn,
    cmap='coolwarm',
    vmin=-1,
    vmax=1
)

plt.title('Stock Return Pearson Correlation')
plt.show()
```

```python
G = nx.Graph()

G.add_nodes_from(correlation_matrix_logreturn.columns)

print("Number of nodes:", G.number_of_nodes())
threshold = 0.5

for i in range(len(correlation_matrix_logreturn.columns)):
    for j in range(i + 1, len(correlation_matrix_logreturn.columns)):

        corr = correlation_matrix_logreturn.iloc[i, j]

        if corr >= threshold:
            stock1 = correlation_matrix_logreturn.columns[i]
            stock2 = correlation_matrix_logreturn.columns[j]

            G.add_edge(stock1, stock2, weight=corr)
```

```python
matched_nodes = [node for node in G.nodes() if node in sector_mapping]
unmatched_nodes = [node for node in G.nodes() if node not in sector_mapping]

print("Total nodes:", G.number_of_nodes())
print("Matched nodes:", len(matched_nodes))
print("Unmatched nodes:", len(unmatched_nodes))

print("\nUnmatched stocks:")
print(unmatched_nodes)
```

```python
nx.set_node_attributes(G, sector_mapping, "sector")
```

```python
sector_assortativity = nx.attribute_assortativity_coefficient(
    G, "sector"
)

print("Sector Assortativity Coefficient:", sector_assortativity)
```

```python
same_sector_edges = 0

for u, v in G.edges():
    if G.nodes[u]["sector"] == G.nodes[v]["sector"]:
        same_sector_edges += 1

total_edges = G.number_of_edges()

p_same = same_sector_edges / total_edges

print("Same-sector edges:", same_sector_edges)
print("Total edges:", total_edges)
print("Proportion of same-sector edges:", p_same)
print("Percentage of same-sector edges:", p_same * 100, "%")
```

```python
import matplotlib.pyplot as plt
import networkx as nx
from matplotlib.patches import Patch

# 1. Extract the giant connected component
giant_nodes = max(nx.connected_components(G), key=len)
G_giant = G.subgraph(giant_nodes).copy()

# 2. Sector colours
unique_sectors = sorted(set(
    nx.get_node_attributes(G_giant, "sector").values()
))

cmap = plt.cm.tab20
sector_colors = {
    sector: cmap(i % 20)
    for i, sector in enumerate(unique_sectors)
}

node_colors = [
    sector_colors[G_giant.nodes[node]["sector"]]
    for node in G_giant.nodes()
]

# 3. Node sizes based on degree within the giant component
giant_degrees = dict(G_giant.degree())

node_sizes = [
    100 + giant_degrees[node] * 20
    for node in G_giant.nodes()
]

# 4. Layout
pos = nx.spring_layout(
    G_giant,
    seed=42,
    k=0.65,
    iterations=500
)

# 5. Plot
fig, ax = plt.subplots(figsize=(18, 14))

nx.draw_networkx_edges(
    G_giant, pos,
    ax=ax,
    alpha=0.12,
    width=0.6,
    edge_color="gray"
)

nx.draw_networkx_nodes(
    G_giant, pos,
    ax=ax,
    node_color=node_colors,
    node_size=node_sizes,
    alpha=0.85
)

# Label the top 10 stocks by degree in giant component
top_nodes = sorted(
    giant_degrees,
    key=giant_degrees.get,
    reverse=True
)[:10]

labels = {
    node: node.replace(".NS.csv", "")
    for node in top_nodes
}

nx.draw_networkx_labels(
    G_giant, pos,
    labels=labels,
    ax=ax,
    font_size=10,
    font_weight="bold"
)

# 6. Legend
legend_handles = [
    Patch(facecolor=sector_colors[s], label=s)
    for s in unique_sectors
]

ax.legend(
    handles=legend_handles,
    title="Industry / Sector",
    loc="center left",
    bbox_to_anchor=(1.02, 0.5),
    fontsize=9
)

ax.set_title(
    "Nifty 200 Stock Correlation Network: Giant Component by Sector (r ≥ 0.5)",
    fontsize=15
)

ax.axis("off")

plt.tight_layout()

plt.savefig(
    "sector_giant_component.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

**D. Threshold Sensitivity Analysis (The "What-If" Study)**

```python
thresholds = [0.5, 0.6, 0.7, 0.8]

graphs = {}

for threshold in thresholds:
    G_temp = nx.Graph()
    G_temp.add_nodes_from(correlation_matrix_logreturn.columns)

    for i in range(len(correlation_matrix_logreturn.columns)):
        for j in range(i + 1, len(correlation_matrix_logreturn.columns)):
            corr = correlation_matrix_logreturn.iloc[i, j]

            if corr >= threshold:
                stock1 = correlation_matrix_logreturn.columns[i]
                stock2 = correlation_matrix_logreturn.columns[j]

                G_temp.add_edge(stock1, stock2, weight=corr)

    graphs[threshold] = G_temp
```

```python
results = []

for threshold, G_temp in graphs.items():

    number_of_edges = G_temp.number_of_edges()

    components = list(nx.connected_components(G_temp))
    giant_component = max(components, key=len)

    giant_size = len(giant_component)

    G_giant = G_temp.subgraph(giant_component)

    avg_path_length = nx.average_shortest_path_length(G_giant)

    results.append({
        "Threshold": threshold,
        "Number of Edges": number_of_edges,
        "Giant Component Size": giant_size,
        "Average Path Length": avg_path_length
    })

sensitivity_df = pd.DataFrame(results)

sensitivity_df
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(
    sensitivity_df["Threshold"],
    sensitivity_df["Number of Edges"],
    marker="o"
)
axes[0].set_title("Number of Edges")
axes[0].set_xlabel("Correlation Threshold")
axes[0].set_ylabel("Edges")

axes[1].plot(
    sensitivity_df["Threshold"],
    sensitivity_df["Giant Component Size"],
    marker="o"
)
axes[1].set_title("Giant Component Size")
axes[1].set_xlabel("Correlation Threshold")
axes[1].set_ylabel("Number of Stocks")

axes[2].plot(
    sensitivity_df["Threshold"],
    sensitivity_df["Average Path Length"],
    marker="o"
)
axes[2].set_title("Average Path Length")
axes[2].set_xlabel("Correlation Threshold")
axes[2].set_ylabel("Path Length")

plt.tight_layout()
plt.show()
```
